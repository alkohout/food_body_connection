# Capstone Project
Our aim is to use machine learning to help identify foods which trigger symptoms such as health or behaviour.
The first phase of the project will be building the machine learning algorithm.
This will help in the correct design and format of the SQL database.
The second phase will involve building an IOS app used to log data and populate the SQL dataset.
The system will be then by tested by populating the database with fake data. 
The populating of the data will have known triggers and health outputs, so that the effectiveness of the model can be tested.
